```python
# -*- coding: utf-8 -*-
"""
Created on Mon Mar 10 12:50:11 2025

@author: Alina
"""

"выводит мощность в виде графика за некоторое время, время "

import pyvisa
import time
import csv
import matplotlib.pyplot as plt

# Создаём объект ResourceManager
rm = pyvisa.ResourceManager()

# Список всех доступных устройств
devices = rm.list_resources()
print(f"Найденные устройства: {devices}")

if not devices:
    print("❌ Нет доступных устройств!")
    exit(1)

device_name = devices[0]  # Выбираем первое устройство

class PMDevicePM100D:
    def __init__(self, device_name=device_name, wl=1555):
        self.rm = pyvisa.ResourceManager()
        self.wavelength = wl
        try:
            self.pm = self.rm.open_resource(device_name)
            print(self.__class__.__name__, "inited!")
            self.set_wl(wl)
        except Exception as e:
            print(f"Ошибка: {e}")
            print("No device found. Devices registered in the system:")
            print(self.rm.list_resources())
            exit(0)

    def __del__(self):
        """Закрываем соединение"""
        if hasattr(self, "pm"):
            try:
                self.pm.close()
            except Exception as e:
                print(f"Error closing PM session: {e}")
        if hasattr(self, "rm"):
            self.rm.close()

    def get_power(self):
        """Запрашивает мощность у прибора"""
        if not hasattr(self, "pm"):
            raise AttributeError("Device was not initialized properly.")

        power = float(self.pm.query("MEASURE:POWER?"))
        time.sleep(0.1)
        return power

    def set_wl(self, wl):
        """Устанавливает длину волны"""
        if hasattr(self, "pm"):
            self.pm.write(f"CORR:WAV {wl}")
            self.wavelength = wl
        return wl

    def get_wl(self):
        """Возвращает текущую длину волны"""
        return self.wavelength

    def record_power(self, duration=10, interval=0.5, filename="C:/Users/Alina/Documents/power_data.csv"):
        """
        Записывает мощность в течение указанного времени и сохраняет в CSV.

        duration: время записи (секунды)
        interval: задержка между измерениями (секунды)
        filename: путь сохранения файла
        """
        print(f"Recording power data for {duration} seconds...")

        start_time = time.time()
        data = []  # Список для хранения данных

        while (time.time() - start_time) < duration:
            t = round(time.time() - start_time, 2)  # Текущее время записи
            power = self.get_power()
            data.append((t, power))
            print(f"Time: {t}s, Power: {power} W")
            time.sleep(interval)  # Ожидание перед следующим измерением

        # Сохранение данных в CSV
        with open(filename, mode="w", newline="") as file:
            writer = csv.writer(file)
            writer.writerow(["Time (s)", "Power (W)"])  # Заголовки
            writer.writerows(data)

        print(f"✅ Data saved to {filename}")
        return data

    def plot_power_data(self, data):
        """Строит график мощности по времени"""
        if not data:
            print("⚠️ Нет данных для построения графика!")
            return

        times, powers = zip(*data)  
        plt.figure(figsize=(8, 5))
        plt.plot(times, powers, marker="o", linestyle="-", label="Power over time")
        plt.xlabel("Time (s)")
        plt.ylabel("Power (W)")
        plt.title(f"Power Measurement at {self.get_wl()} nm")
        plt.legend()
        plt.grid(True)
        plt.show()


# === Запуск измерения ===
if __name__ == "__main__":
    pm = PMDevicePM100D()
    data = pm.record_power(duration=30, interval=1)  # время и шаг
    pm.plot_power_data(data)
```

```python
import pyvisa as visa
import time
import numpy as np
class PMDevicePM100D():
    def __init__(self, device_name = 'USB0::0x1313::0x8075::P5001126::INSTR', wl = 1064):
        self.rm = visa.ResourceManager()
        try:
            self.pm = self.rm.open_resource(device_name)
            print(self.__class__.__name__, 'inited!')
            self.pm.write('CORR:WAV {}'.format(wl))
        except Exception as e:
            print(e)
            print('No dece found. Devices registered in the system:')
            self.rm.list_resources()
            exit(0)


    def disconnect(self):
        """ Отключить соединение"""
        self.pm.control_ren(False)
        self.pm.close()
        self.rm.close()
        print('Power meter is disconnect')

    def get_power(self):
        power = float(self.pm.query('measure:power?'))
        time.sleep(0.1)
        return(power)

    def set_wl(self, wl = 1070):
        self.pm.write('CORR:WAV {}'.format(wl))
        return wl
    
def measure_average_power(pm_device,duration,aver_point ):
    
    power_values = []
    for _ in range(aver_point):  
        power = pm_device.get_power()
        power_values.append(power)
        time.sleep(duration / aver_point)
    
    # Вычисляем среднее и переводим в мВт
    avg_pm_power = np.mean(power_values) * 1000 
    return round(avg_pm_power, 4)


if __name__ == "__main__":
    pm_device=PMDevicePM100D()
    pm_power = measure_average_power(pm_device=pm_device,duration=3,aver_point=5)
    print('Сurrent pm power =',pm_power,'mW')
```
